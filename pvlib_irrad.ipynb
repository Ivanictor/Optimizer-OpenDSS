Código PVLIB Python

```python
import pvlib

bus_list_df = buses_df[buses_df["name"].isin(bus_list)]

coordinates = []
for _, row in bus_list_df.iterrows():
    coord = (row["x"], row["y"])
    coordinates.append(coord)

start_year = 2019
end_year = 2023

period = end_year-start_year+1
irrad = []
t = 1
for location in coordinates:
    longitude, latitude = location
    data = pvlib.iotools.get_pvgis_hourly(latitude, longitude, start=start_year, end=end_year, components=False,timeout=300)[0]
    print('Baixou %i' % t, end = '\r')
    t+=1
    irrad.append(data)
```

Organização dos Dados

```python
import pandas as pd

index_time = pd.date_range('2010-01-01', '2011-01-01', freq='1h', inclusive='left')
new_index=[]
for i in index_time:
    time = i.strftime('%d-%m %H:%M:%S')
    new_index.append(time)

all_hourly_means = []
h=1
for i in irrad:
    df_5year = pd.DataFrame()
    
    df_5year['2019'] = i.iloc[0:8760]["poa_global"].values
    df_5year['2020'] = pd.concat([i.iloc[8760:10176]["poa_global"], i.iloc[10200:17544]["poa_global"]]).values #Remove o dia extra do ano bissexto
    df_5year['2021'] = i.iloc[17544:26304]["poa_global"].values
    df_5year['2022'] = i.iloc[26304:35064]["poa_global"].values
    df_5year['2023'] = i.iloc[35064:43824]["poa_global"].values
       
    df_5year['mean'] = df_5year.mean(axis=1)
    df_5year.index = new_index
    df_5year.to_csv('./irrad_points/point-%i.csv' % h)
    h+=1
    df_5year.index = pd.date_range('2010-01-01', '2011-01-01', freq='1h', inclusive='left')
    df_5year['hour'] = df_5year.index.hour
    
    hourly_means = df_5year.groupby('hour')['mean'].mean()
    
    all_hourly_means.append(hourly_means)
```

Salvando os dados de irradiância

```python
five_year = []
for t in range(60):
    df = pd.read_csv('./irrad_points/point-%i.csv' % (t+1), index_col=0)
    five_year.append(df)
```

Criação do DataFrame (salvo em CSV)

```python
hourly_means_df = pd.DataFrame(all_hourly_means)
for hour in range(24):
    col_name = f'irrad_{hour}h'
    bus_list_df[col_name] = hourly_means_df.iloc[:, hour].values

display(bus_list_df)
bus_list_df.to_csv("./irrad_points/buses_with_irradiance.csv")
```

Testes com o DataFrame dos Grafos

```python
graph_df = dss_tools.model.graph_df
buses_df = dss_tools.model.buses_df
count_bus12, count_bus21, count_bus_equal = 0, 0, 0

for _, row in graph_df.iterrows():
    bus1 = row["bus1"]
    bus2 = row["bus2"]

    distance_1 = buses_df.query(f"name == '{bus1}'")["distance"].iloc[0]
    distance_2 = buses_df.query(f"name == '{bus2}'")["distance"].iloc[0]

    if distance_1 > distance_2:
        count_bus12 += 1

    elif distance_1 < distance_2:
        count_bus21 += 1

    elif distance_1 == distance_2:
        count_bus_equal += 1
        print(f"Bus1: {row["bus1"]}, Bus2: {row["bus2"]}")


        

print("Contagem de grafos com bus1 mais distante que bus2: ", count_bus12)
print("Contagem de grafos com bus1 mais próximo que bus2: ", count_bus21)
print("Contagem de grafos com bus1 igualmente distante do source em relação a bus2: ", count_bus_equal)
print("Total: ", len(graph_df))
    

```
