```python
import pandas as pd

loads_df["kw"] = pd.to_numeric(loads_df["kw"], errors="coerce")

loads_df["kv"] = pd.to_numeric(loads_df["kv"], errors="coerce")
```

```python
display(loads_df["kw"].describe())

display(loads_df["kv"].value_counts().reset_index())

loads_df = loads_df.sort_values(by="kw", ascending=False)
```

```python
lines_df = dss_tools.model.lines_df

contador_bt = 0
contador_mt = 0
for _, row in lines_df.iterrows():
    if "bt" in row["name"]:
        contador_bt += 1

    elif "mt" in row["name"]:
        contador_mt += 1

print(f"Total Linhas BT: {contador_bt}")
print(f"Total Linhas MT: {contador_mt}")

display(lines_df)
```
